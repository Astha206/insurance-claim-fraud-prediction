```python
categorical_cols = [
    "Sex",
    "Fault",
    "AccidentArea",
    "VehicleCategory",
    "VehiclePrice",
    "PolicyType",
    "BasePolicy",
    "AgentType",
    "WitnessPresent",
    "NumberOfSuppliments",
    "AddressChange_Claim",
    "NumberOfCars"
]
```

```python
for col in categorical_cols:
    print("\n---", col, "---")

    print("Group size:")
    print(df[col].value_counts())

    print("\nGroup percentage:")
    print(df[col].value_counts(normalize=True) * 100)

    print("\nFraud rate:")
    print(df.groupby(col)["FraudFound_P"].mean() * 100)
```
